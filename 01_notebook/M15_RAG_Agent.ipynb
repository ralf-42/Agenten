# 1 | Übersicht

In *RAG-Chain mit LangChain* wurde eine **RAG-Chain** gebaut: deterministisch, mit linearem Ablauf.  
Jetzt folgt der nächste Schritt: **RAG als Tool für einen Agenten**.

**Von der Chain zum Agent**

| Merkmal | RAG-Chain (M14) | RAG-Agent (M15) |
|---------|----------------|----------------|
| Ablauf | Immer: Retrieve → Generate | Nur bei Bedarf: Agent entscheidet |
| RAG-Nutzung | Immer aktiv | Tool, das der Agent auswählt |
| Weiteres Wissen | Nur Vektordatenbank | Modellwissen + RAG + weitere Tools |
| Flexibilität | Niedrig | Hoch |
| Komplexität | Gering | Mittel |

**Was entsteht**

Ein Agenten-System mit **RAG als einem von mehreren Tools**:
- `suche_wissensdatenbank` – RAG-Tool für belegbare Fragen zum Meeting-Briefing-Korpus
- `erklaere_konzept` – Direktes Modellwissen für allgemeine Erklärungen
- `schaetze_tokens` – Einfaches Berechnungstool für Kontextkosten


# 2 | RAG als Tool
---

Ein **RAG-Tool** ist eine normale LangChain-Tool-Funktion, die intern ChromaDB abfragt. Der Agent entscheidet, ob diese Quelle nötig ist.


Der Index-Aufbau ist hier Voraussetzung, nicht das Lernziel. Der Fokus liegt auf Tool-Definition, Tool-Wahl und Quellen-Grenzen des Agenten.



<p><font color='black' size="5">
Design-Prinzipien für RAG-Tools
</font></p>


| Prinzip | Beschreibung |
|---------|-------------|
| **Klarer Name** | Name sagt dem Agenten, wann das Tool zu nutzen ist |
| **Präzise Beschreibung** | Docstring erklärt genau, welche Fragen das Tool beantwortet |
| **Fehlerbehandlung** | Gibt lesbaren Text zurück, kein `raise` |
| **Formatierte Ausgabe** | Gibt Quellen + Antwort zurück |

```python
from pathlib import Path

from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_chroma import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.documents import Document
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
from langchain_openai import OpenAIEmbeddings

from genai_lib.model_config import WORKER
# LLM und Embeddings
llm = init_chat_model(WORKER)
embeddings = OpenAIEmbeddings(model=EMBEDDINGS)
```

```python
def load_pdf_documents(directory_path: str):
    """Lädt alle PDF-Seiten aus einem Verzeichnis."""
    documents = []
    for pdf_path in sorted(Path(directory_path).glob("*.pdf")):
        try:
            loader = PyPDFLoader(str(pdf_path))
            documents.extend(loader.load())
        except Exception as exc:
            print(f"  ⚠ {pdf_path.name} übersprungen: {exc}")
    return documents


def split_documents_by_characters(documents, chunk_size=300, chunk_overlap=30):
    """Teilt Dokumentseiten in überlappende Zeichen-Chunks."""
    chunks = []
    step = max(1, chunk_size - chunk_overlap)
    for doc in documents:
        text = doc.page_content or ""
        for chunk_index, start in enumerate(range(0, len(text), step)):
            chunk_text = text[start:start + chunk_size].strip()
            if not chunk_text:
                continue
            metadata = dict(doc.metadata)
            metadata["chunk_index"] = chunk_index
            chunks.append(Document(page_content=chunk_text, metadata=metadata))
    return chunks


# Dokumente laden und splitten
dokumente_raw = load_pdf_documents(KORPUS_TARGET)
chunks = split_documents_by_characters(
    dokumente_raw,
    chunk_size=300,
    chunk_overlap=30,
)

# ChromaDB aufbauen (persistent, auf Google Drive) — mit Batching für große Korpora
BATCH_SIZE = 500
vectorstore = Chroma(
    collection_name="meeting_briefing_korpus_m14",
    embedding_function=embeddings,
    persist_directory="/content/drive/MyDrive/Agenten/02_daten/05_sonstiges/chroma_briefing",
)

count_vorher = vectorstore._collection.count()
if count_vorher == 0:
    for start in range(0, len(chunks), BATCH_SIZE):
        batch = chunks[start:start + BATCH_SIZE]
        vectorstore.add_documents(batch)
        ende = min(start + BATCH_SIZE, len(chunks))
        print(f"  Batch {start // BATCH_SIZE + 1}: {ende}/{len(chunks)} Chunks indexiert")
else:
    print(f"Collection hat bereits {count_vorher} Einträge — kein erneutes Indexieren")

print(f"Geladene PDF-Seiten: {len(dokumente_raw)}")
print(f"Chunks verfügbar:    {len(chunks)}")
print(f"Collection-Größe:    {vectorstore._collection.count()}")
```

# 3 | RAG-Agent bauen
---



```python
from genai_lib.model_config import WORKER
@tool
def suche_wissensdatenbank(frage: str) -> str:
    """Durchsucht den Meeting- & Projekt-Korpus zu Entscheidungen, Risiken und offenen Punkten
    sowie die Fachartikel zu RAG, Retrieval, Evaluation und Agenten.
    Geeignet für belegbare Fragen zu Projekt Kompass und Fragen nach Quellen.
    """
    try:
        ergebnisse = vectorstore.similarity_search_with_score(frage, k=3)
        if not ergebnisse:
            return "Nicht im Korpus. Keine relevanten Informationen gefunden."

        antwort_teile = []
        for doc, score in ergebnisse:
            quelle = doc.metadata.get("source", "unbekannt").split("/")[-1]
            ausschnitt = doc.page_content.replace("\n", " ")[:700]
            antwort_teile.append(
                f"Quelle: {quelle}\n"
                f"Score: {score:.4f}\n"
                f"Ausschnitt: {ausschnitt}"
            )

        return "\n\n---\n\n".join(antwort_teile)
    except Exception as exc:
        return f"Fehler bei der Datenbanksuche: {exc}"


@tool
def erklaere_konzept(konzept: str) -> str:
    """Erklärt ein allgemeines KI- oder Informatik-Konzept kurz und verständlich.
    Geeignet für Grundlagenfragen ohne zwingenden Bezug zum Meeting-Briefing-Korpus.
    """
    erklaer_llm = init_chat_model(WORKER)
    prompt = f"Erkläre das Konzept '{konzept}' in 2-3 Sätzen für einen Einstiegskurs."
    return erklaer_llm.invoke(prompt).content


@tool
def schaetze_tokens(text: str, modell: str = "gpt-5.4-mini") -> str:
    """Schätzt Tokenzahl und grobe Eingabekosten für einen Text."""
    zeichen = len(text)
    tokens_grob = zeichen // 3
    kosten_pro_1000 = 0.005
    kosten = (tokens_grob / 1000) * kosten_pro_1000
    return (
        f"Text: {zeichen} Zeichen\n"
        f"Geschätzte Tokens: ~{tokens_grob} (Faustregel: 3 Zeichen/Token)\n"
        f"Geschätzte Kosten ({modell}): ~${kosten:.6f}"
    )


tools = [suche_wissensdatenbank, erklaere_konzept, schaetze_tokens]
print(f"Tools definiert: {[t.name for t in tools]}")
```

```python
# RAG-Agent erstellen
system_prompt = load_prompt("https://github.com/ralf-42/Agenten/blob/main/05_prompt/m14_rag_agent_system_prompt.md", mode="S")

rag_agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt,
    debug=False
)

print("RAG-Agent erstellt")
print(f"Tools: {[t.name for t in tools]}")
```

# 4 | Wann RAG, wann eigenes Wissen?
---



Der Agent entscheidet selbst, welches Tool er nutzt – aber wie trifft er diese Entscheidung?

**Entscheidungslogik**

Der Agent wählt das Tool basierend auf **drei Faktoren**:

1. **Tool-Name** – Sprechende Namen helfen: `suche_wissensdatenbank` vs. `erklaere_konzept`
2. **Docstring** – Präzise Beschreibung, wann das Tool einzusetzen ist
3. **Kontext der Frage** – Spezifisch (→ RAG) vs. allgemein (→ eigenes Wissen)

```python
#@markdown   <p><font size="4" color='green'>  flowchart</font> </br></p>

diagram = '''
%%{init: {'theme':'forest'}}%%
flowchart TD
    FRAGE["❓ Nutzerfrage"]

    FRAGE --> SPEZ{{"Research-Frage?
(RAG, Retrieval,
Evaluation, Agenten)"}}
    SPEZ -->|"Ja"| RAG["📚 suche_wissensdatenbank()
ChromaDB Retrieval"]
    SPEZ -->|"Nein"| ALLG{{"Allgemeines
Konzept?"}}

    ALLG -->|"Ja"| LLM["🧠 erklaere_konzept()
Modellwissen"]
    ALLG -->|"Nein"| CALC{{"Berechnung
nötig?"}}

    CALC -->|"Ja"| TOK["🔢 schaetze_tokens()
Faustregel"]
    CALC -->|"Nein"| DIRECT["💬 Direkte Antwort"]

    RAG --> ANS["Antwort"]
    LLM --> ANS
    TOK --> ANS
    DIRECT --> ANS
'''

mermaid(diagram, width=850)
```

```python
# Test: Agent zeigt unterschiedliche Tool-Wahl
test_fragen = [
    # Erwartet: suche_wissensdatenbank
    "Wie unterscheidet sich naive RAG von advanced RAG?",
    # Erwartet: suche_wissensdatenbank
    "Warum kann irrelevanter Kontext bei RAG problematisch sein?",
    # Erwartet: erklaere_konzept oder direkte Antwort
    "Was ist ein Embedding-Vektor in der KI?",
    # Erwartet: schaetze_tokens
    "Schätze die Tokenzahl für diesen kurzen Beispieltext zur RAG-Evaluation.",
]

run_cfg = {"run_name": "M15-RAG-Agent-Tool-Wahl", "tags": ["rag-agent", "m15", "meeting-briefing-korpus"]}

toolwahl_ergebnisse = []
for frage in test_fragen:
    print(f"Frage: {frage}")
    ergebnis = rag_agent.invoke(
        {"messages": [HumanMessage(content=frage)]},
        config=run_cfg,
    )
    letzte_antwort = ergebnis["messages"][-1].content
    tool_calls = [m for m in ergebnis["messages"] if getattr(m, "type", None) == "tool"]
    toolwahl_ergebnisse.append({"frage": frage, "tool_calls": len(tool_calls), "antwort": letzte_antwort})
    print(f"Tool-Aufrufe: {len(tool_calls)}")
    print(f"Antwort: {letzte_antwort[:300]}...")
    print("-" * 60)
    print()
```

# 5 | End-to-End Testing
---



Ein gutes RAG-System braucht systematische Tests.  




| Testtyp | Was wird geprüft? | Methode |
|---------|-------------------|--------|
| **Retrieval-Test** | Werden relevante Dokumente gefunden? | Manuell prüfen |
| **Antwortqualität** | Sind die Antworten korrekt? | Expected vs. Actual |
| **Tool-Wahl** | Wählt der Agent das richtige Tool? | Trace analysieren |

```python
# End-to-End Test: vollständiger Durchlauf
test_szenarien = [
    {
        "frage": "Wie unterscheidet sich naive RAG von advanced RAG?",
        "keywords": ["naive", "advanced", "RAG"],
    },
    {
        "frage": "Warum ist Evaluation bei RAG-Systemen wichtig?",
        "keywords": ["Evaluation", "RAG"],
    },
]

print("=== End-to-End Test ===\n")
run_cfg = {"run_name": "M15-E2E-Test", "tags": ["test", "m15", "meeting-briefing-korpus"]}

e2e_ergebnisse = []
for i, szenario in enumerate(test_szenarien):
    frage = szenario["frage"]
    print(f"Test {i+1}: {frage}")

    ergebnis = rag_agent.invoke(
        {"messages": [HumanMessage(content=frage)]},
        config=run_cfg,
    )

    antwort = ergebnis["messages"][-1].content
    treffer = sum(
        1 for wort in szenario["keywords"]
        if wort.lower() in antwort.lower()
    )
    gesamt = len(szenario["keywords"])
    status = "PASS" if treffer >= max(1, gesamt // 2) else "FAIL"
    e2e_ergebnisse.append({"frage": frage, "treffer": treffer, "gesamt": gesamt, "status": status})

    print(f"  Antwort: {antwort[:220]}...")
    print(f"  Schlüsselwörter: {treffer}/{gesamt} gefunden")
    print(f"  Status: {status}")
    print()
```

**Zusammenfassung: Wann was?**

| Kriterium | RAG-Chain | RAG-Agent |
|-----------|-----------|-----------|
| Einfache Q&A | Ideal | Overhead |
| Gemischte Fragen | Nur RAG | Flexibel |
| Mehrere Tools | Komplex | Geeignet |
| Nachvollziehbarkeit | Deterministisch | Trace nötig |
| Einstieg | Empfohlen | Nachgelagert |

**Empfehlung:** Mit RAG-Chain starten, bei Bedarf zum RAG-Agenten ausbauen.
