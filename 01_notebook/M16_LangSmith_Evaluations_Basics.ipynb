# 3 | Research-Index für Evaluation
---


M16 weicht bewusst von der M13-Baseline ab: Für Evaluation sind satz-/absatznahe Chunks hilfreicher, weil Scorer einzelne Aussagen und Quellenbindung prüfen sollen.

Der Evaluationslauf nutzt denselben stabilen PDF-RAG-Pfad wie M13 bis M15: `PyPDFLoader`, Chroma-Batching und eine einfache lokale Chunking-Funktion. Für die Qualitätsbewertung wird absatz- und satznäher gesplittet, damit zentrale Fachbegriffe seltener an festen Zeichengrenzen getrennt werden.

📦 **Hinweis zum Modul `genai_lib.briefing_rag`:** Ab M15 wird die produktive Chroma-Kette in dieses zentrale, auf Google Drive persistierte Modul ausgelagert. M16 bleibt davon bewusst unabhängig: `split_text_to_units`/`split_documents_by_sections` splitten satz-/absatzbasiert statt zeichenbasiert — eine strukturell andere Methode als die Modul-Funktion, die aktuell keinen zweiten Anwendungsfall hat. Eine Modul-Funktion nur für diesen einen Aufrufer wäre verfrühte Abstraktion. Die eigene, lokale Eval-Collection (`research_eval_m15_sentence`) bleibt deshalb unverändert bestehen.

```python
from langchain.chat_models import init_chat_model
from langchain_chroma import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.documents import Document
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_openai import OpenAIEmbeddings

import re

from genai_lib.model_config import JUDGE
llm = init_chat_model(JUDGE)
embeddings = OpenAIEmbeddings(model=EMBEDDINGS)


def load_pdf_documents(directory_path: str):
    documents = []
    for pdf_path in sorted(Path(directory_path).glob("*.pdf")):
        try:
            loader = PyPDFLoader(str(pdf_path))
            documents.extend(loader.load())
        except Exception as exc:
            print(f"  ⚠ {pdf_path.name} übersprungen: {exc}")
    return documents


def split_text_to_units(text: str):
    paragraphs = [p.strip() for p in re.split(r"\n\s*\n", text or "") if p.strip()]
    units = []
    for paragraph in paragraphs:
        if len(paragraph) <= 700:
            units.append(paragraph)
            continue
        sentences = re.split(r"(?<=[.!?])\s+", paragraph)
        units.extend(s.strip() for s in sentences if s.strip())
    return units


def split_documents_by_sections(documents, chunk_size=700, chunk_overlap_units=1):
    chunks = []
    for doc in documents:
        units = split_text_to_units(doc.page_content)
        current = []
        current_len = 0
        chunk_index = 0
        for unit in units:
            unit_len = len(unit)
            if current and current_len + unit_len + 1 > chunk_size:
                chunk_text = " ".join(current).strip()
                metadata = dict(doc.metadata)
                metadata["chunk_index"] = chunk_index
                metadata["chunking"] = "paragraph_sentence"
                chunks.append(Document(page_content=chunk_text, metadata=metadata))
                chunk_index += 1
                current = current[-chunk_overlap_units:] if chunk_overlap_units else []
                current_len = sum(len(item) for item in current)
            current.append(unit)
            current_len += unit_len + 1
        if current:
            chunk_text = " ".join(current).strip()
            metadata = dict(doc.metadata)
            metadata["chunk_index"] = chunk_index
            metadata["chunking"] = "paragraph_sentence"
            chunks.append(Document(page_content=chunk_text, metadata=metadata))
    return chunks


dokumente_raw = load_pdf_documents(KORPUS_TARGET)
chunks = split_documents_by_sections(dokumente_raw)

vectorstore = Chroma(
    collection_name="research_eval_m15_sentence",
    embedding_function=embeddings,
    persist_directory="chroma_m15_sentence",
)

BATCH_SIZE = 500
count_vorher = vectorstore._collection.count()
if count_vorher == 0:
    for start in range(0, len(chunks), BATCH_SIZE):
        batch = chunks[start:start + BATCH_SIZE]
        vectorstore.add_documents(batch)
        ende = min(start + BATCH_SIZE, len(chunks))
        print(f"  Batch {start // BATCH_SIZE + 1}: {ende}/{len(chunks)} Chunks indexiert")
else:
    print(f"Collection hat bereits {count_vorher} Einträge — kein erneutes Indexieren")

retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
print(f"PDF-Seiten:       {len(dokumente_raw)}")
print(f"Chunks verfügbar: {len(chunks)}")
print(f"Collection-Größe: {vectorstore._collection.count()}")
```

# 4 | Retrieval-Bewertung
---


```python
def normalisiere_text(text: str) -> str:
    return (text or "").lower()


def quellen_name(doc) -> str:
    return doc.metadata.get("source", "?").split("/")[-1]


def begriffgruppen(fall: dict):
    gruppen = fall.get("erwartete_begriffgruppen")
    if gruppen:
        return gruppen
    return [[begriff] for begriff in fall.get("erwartete_begriffe", [])]


def bewerte_begriffgruppen(text: str, fall: dict):
    text_norm = normalisiere_text(text)
    gruppen = begriffgruppen(fall)
    gruppen_treffer = []
    for gruppe in gruppen:
        treffer = [begriff for begriff in gruppe if normalisiere_text(begriff) in text_norm]
        if treffer:
            gruppen_treffer.append(treffer[0])
    score = len(gruppen_treffer) / len(gruppen) if gruppen else 1.0
    return score, gruppen_treffer


def bewerte_retrieval(fall: dict, k: int = 3) -> dict:
    docs = vectorstore.similarity_search(fall["frage"], k=k)
    quellen = [quellen_name(doc) for doc in docs]
    text = "\n".join(doc.page_content for doc in docs)

    erwartete_quelle = fall.get("erwartete_quelle")
    quelle_getroffen = erwartete_quelle is None or erwartete_quelle in quellen

    begriff_score, begriff_treffer = bewerte_begriffgruppen(text, fall)

    return {
        "id": fall["id"],
        "set": fall["set"],
        "typ": fall["typ"],
        "frage": fall["frage"],
        "erwartete_quelle": erwartete_quelle,
        "quellen": quellen,
        "quelle_getroffen": quelle_getroffen,
        "begriff_score": round(begriff_score, 3),
        "begriff_treffer": begriff_treffer,
    }


retrieval_ergebnisse = [bewerte_retrieval(fall) for fall in alle_eval_faelle]
quelle_accuracy = sum(r["quelle_getroffen"] for r in retrieval_ergebnisse) / len(retrieval_ergebnisse)
begriff_mittel = sum(r["begriff_score"] for r in retrieval_ergebnisse) / len(retrieval_ergebnisse)

print(f"Quellen-Trefferquote: {quelle_accuracy:.2%}")
print(f"Begriff-Abdeckung:    {begriff_mittel:.2%}")
```

```python
rows = [
    "| ID | Set | Quelle getroffen | Begriff-Score | Quellen |",
    "|---|---|---:|---:|---|",
]
for r in retrieval_ergebnisse[:12]:
    rows.append(
        f"| {r['id']} | {r['set']} | {int(r['quelle_getroffen'])} | "
        f"{r['begriff_score']:.2f} | {', '.join(r['quellen'][:3])} |"
    )

mprint("## Retrieval-Baseline\n\n" + "\n".join(rows))
```

# 5 | Einfache Antwortbewertung
---


```python
rag_prompt = PromptTemplate.from_template(
    "Antwort ausschließlich auf Basis des Kontexts formulieren. "
    "Wenn der Kontext nicht reicht: Nicht im Korpus.\n\n"
    "Kontext:\n{context}\n\n"
    "Frage: {question}\n\n"
    "Antwort mit Quellenhinweis:"
)


def format_docs(docs):
    teile = []
    for doc in docs:
        quelle = quellen_name(doc)
        teile.append(f"Quelle: {quelle}\n{doc.page_content}")
    return "\n\n---\n\n".join(teile)


rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | rag_prompt
    | llm
    | StrOutputParser()
)


def beantworte_eval_fall(fall: dict) -> dict:
    docs = retriever.invoke(fall["frage"], config=run_cfg)
    antwort = rag_chain.invoke(fall["frage"], config=run_cfg)
    return {
        "id": fall["id"],
        "frage": fall["frage"],
        "antwort": antwort,
        "quellen": [quellen_name(doc) for doc in docs],
    }

print("RAG-Evaluationsfunktion bereit")
```

```python
def bewerte_antwort(fall: dict, antwort: str) -> dict:
    begriff_score, begriff_treffer = bewerte_begriffgruppen(antwort, fall)

    erwartet_nicht_im_korpus = fall.get("erwartete_quelle") is None
    antwort_norm = normalisiere_text(antwort)
    nicht_im_korpus_ok = (
        not erwartet_nicht_im_korpus
        or "nicht im korpus" in antwort_norm
        or "keine" in antwort_norm
    )

    return {
        "id": fall["id"],
        "antwort_begriff_score": round(begriff_score, 3),
        "nicht_im_korpus_ok": nicht_im_korpus_ok,
        "antwort_begriff_treffer": begriff_treffer,
    }


# Kleine Stichprobe für Kostenkontrolle im Kurslauf.
eval_stichprobe = haupt_eval[:5] + edge_eval[:3]
eval_stichprobe = [
    {**fall, "set": "main"} if fall["id"].startswith("q") else {**fall, "set": "edge"}
    for fall in eval_stichprobe
]

antwort_ergebnisse = []
for fall in eval_stichprobe:
    output = beantworte_eval_fall(fall)
    scores = bewerte_antwort(fall, output["antwort"])
    antwort_ergebnisse.append({**output, **scores})
    print(f"{fall['id']}: Antwort-Score {scores['antwort_begriff_score']:.2f}")

print(f"Antwort-Stichprobe: {len(antwort_ergebnisse)} Fälle")
```

```python
rows = [
    "| ID | Antwort-Begriffe | Nicht-im-Korpus ok | Quellen | Antwortauszug |",
    "|---|---:|---:|---|---|",
]
for r in antwort_ergebnisse:
    auszug = r["antwort"][:140].replace("\n", " ")
    rows.append(
        f"| {r['id']} | {r['antwort_begriff_score']:.2f} | "
        f"{int(r['nicht_im_korpus_ok'])} | {', '.join(r['quellen'][:3])} | {auszug}... |"
    )

mprint("## Antwort-Baseline\n\n" + "\n".join(rows))
```

# 6 | LangSmith-Dataset vorbereiten
---


Das lokale Scoring reicht für den ersten Umbau. Für spätere Regressionstests kann dasselbe Eval-Set als LangSmith-Dataset angelegt und mit `evaluate()` ausgeführt werden.


```python
from langsmith import Client

client = Client()
DATASET_NAME = "M16-Research-Eval-v1"

langsmith_examples = [
    {
        "inputs": {"frage": fall["frage"], "id": fall["id"], "set": fall["set"]},
        "outputs": {
            "antwort": fall["erwartete_antwort_kurz"],
            "erwartete_quelle": fall.get("erwartete_quelle"),
            "erwartete_begriffe": fall.get("erwartete_begriffe", []),
            "erwartete_begriffgruppen": fall.get("erwartete_begriffgruppen"),
        },
    }
    for fall in alle_eval_faelle
]

vorhandene = [ds.name for ds in client.list_datasets()]
if DATASET_NAME in vorhandene:
    dataset = next(ds for ds in client.list_datasets() if ds.name == DATASET_NAME)
    print(f"Dataset wiedergefunden: {DATASET_NAME}")
else:
    dataset = client.create_dataset(
        dataset_name=DATASET_NAME,
        description="Meeting- & Research-Briefing-Agent Eval-Set mit Hauptfragen und Edge-Cases",
    )
    for ex in langsmith_examples:
        client.create_example(
            dataset_id=dataset.id,
            inputs=ex["inputs"],
            outputs=ex["outputs"],
        )
    print(f"Dataset erstellt: {DATASET_NAME}")

mein_dataset_name = DATASET_NAME
examples_in_dataset = list(client.list_examples(dataset_id=dataset.id))
print(f"Examples im Dataset: {len(examples_in_dataset)}")
print("Hinweis: Bei bestehendem Dataset q04 in LangSmith ggf. neu anlegen, damit Alias-Gruppen übernommen werden.")
```

```python
from langsmith.evaluation import evaluate


def research_target(inputs: dict) -> dict:
    fall = {"frage": inputs["frage"]}
    docs = retriever.invoke(fall["frage"], config=run_cfg)
    antwort = rag_chain.invoke(fall["frage"], config=run_cfg)
    return {
        "antwort": antwort,
        "quellen": [quellen_name(doc) for doc in docs],
    }


def quellen_evaluator(inputs: dict, outputs: dict, reference_outputs: dict) -> dict:
    erwartete_quelle = reference_outputs.get("erwartete_quelle")
    quellen = outputs.get("quellen", [])
    score = 1.0 if erwartete_quelle is None or erwartete_quelle in quellen else 0.0
    return {"key": "quelle_top3", "score": score}


def begriff_evaluator(inputs: dict, outputs: dict, reference_outputs: dict) -> dict:
    antwort = outputs.get("antwort", "")
    gruppen = reference_outputs.get("erwartete_begriffgruppen")
    if gruppen:
        fall = {"erwartete_begriffgruppen": gruppen}
    else:
        fall = {"erwartete_begriffe": reference_outputs.get("erwartete_begriffe", [])}
    score, _ = bewerte_begriffgruppen(antwort, fall)
    return {"key": "begriff_abdeckung", "score": round(score, 3)}


def nicht_im_korpus_evaluator(inputs: dict, outputs: dict, reference_outputs: dict) -> dict:
    erwartete_quelle = reference_outputs.get("erwartete_quelle")
    antwort = normalisiere_text(outputs.get("antwort", ""))
    if erwartete_quelle is not None:
        return {"key": "nicht_im_korpus", "score": 1.0}
    score = 1.0 if "nicht im korpus" in antwort else 0.0
    return {"key": "nicht_im_korpus", "score": score}


# Voller LangSmith-Lauf ist bewusst vorbereitet, aber für Kostenkontrolle nicht automatisch nötig.
# meine_eval_results = evaluate(
#     research_target,
#     data=mein_dataset_name,
#     evaluators=[quellen_evaluator, begriff_evaluator, nicht_im_korpus_evaluator],
#     experiment_prefix="M16-Research-Baseline",
#     max_concurrency=1,
# )

print("LangSmith-Evaluation vorbereitet")
print("Zum Ausführen Kommentar vor `meine_eval_results = evaluate(...)` entfernen.")
```
